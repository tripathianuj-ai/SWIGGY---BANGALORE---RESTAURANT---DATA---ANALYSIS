```python
# TASK 3 - FIND PARTICULAR LOCATION WISE RATING & COST FOR TWO PEOPLE

swiggy_data_koramangala['Rating']
```

```python
# Distribution of Rating in Koramangala

sns.histplot(swiggy_data_koramangala,x= swiggy_data_koramangala['Rating'],bins=10,color='purple',zorder=2)

plt.title('DISTRIBUTION OF  KORAMANGALA RATING')

plt.grid()

plt.show()
```

```python
sns.histplot(swiggy_data_koramangala,x= swiggy_data_koramangala['Cost_for_Two'],bins=15,color='yellow',zorder=2)

plt.title('DISTRIBUTION OF COST FOR TWO PEOPLE IN KORAMANGALA')

plt.grid()

plt.show()
```

CONCLUSION FOR KORAMANGALA : 

MAX RATING IS  4.1 TO 4.3  AND COST FOR TWO PEOPLE ARE  200 TO 360  & MAX GOES UPTO  600

```python
# Distribution of Rating in BTM

sns.histplot(swiggy_data_BTM,x= swiggy_data_BTM['Rating'],bins=10,color='cyan',zorder=2)

plt.title('DISTRIBUTION OF BTM RATING')

plt.grid()

plt.show()
```

```python
sns.histplot(swiggy_data_BTM,x= swiggy_data_BTM['Cost_for_Two'],bins=10,color='orange',zorder=2)

plt.title('DISTRIBUTION OF COST FOR TWO PEOPLE IN BTM')

plt.grid()

plt.show()
```

CONCLUSION FOR BTM : 

MAX RATING IS  4.1 TO 4.2 AND COST FOR TWO PEOPLE ARE  250 TO 300  & MAX GOES UPTO  600

```python
# Distribution of Rating in HSR

sns.histplot(swiggy_data_HSR,x= swiggy_data_HSR['Rating'],bins=10,color='pink',zorder=2)

plt.title('DISTRIBUTION OF HSR RATING')

plt.grid()

plt.show()
```

```python
sns.histplot(swiggy_data_HSR,x= swiggy_data_HSR['Cost_for_Two'],bins=10,color='green',zorder=2)

plt.title('DISTRIBUTION OF COST FOR TWO PEOPLE IN HSR')

plt.grid()

plt.show()
```

CONCLUSION FOR HSR : 

MAX RATING IS  4 & ABOVE  AND COST FOR TWO PEOPLE ARE  350 TO 400  & MAX GOES UPTO  800

```python
# TASK 4 - HIGHEST RATING RESTAURANT NAME

high_rated_restaurant = swiggy_data[swiggy_data['Rating']>=4.0]

```

```python
high_rated_restaurant
```

```python
high_rated_restaurant.max()
```

```python
high_rated_restaurant.min()
```

CONCLUSION:

HIGHEST RATING RESTAURANT NAME IS CALICUT CAFE RESTAURANT & RATING IS 4.8

```python
# TASK 5  -  AFFORDABLE RESTAURANTS  WITH HIGHEST RATING 

affordable_restaurant = swiggy_data[(swiggy_data['Rating']>=4.0) & (swiggy_data['Cost_for_Two']<500)]
affordable_restaurant 



```

```python
affordable_restaurant.sort_values(by ='Rating',ascending = False)
```

CONCLUSION :

 AFFORDABLE RESTAURANT IS KHICHDI EXPERIMENT WITH HIGHEST RATING 4.8

```python
# PLOTLY 

import plotly.express as px 
fig = px.scatter(high_rated_restaurant,
                 x='Rating',
                 y='Cost_for_Two',
                 color='Rating',
                 size='Cost_for_Two')

fig.update_layout(template = "plotly_dark",title="HIGH RATED RESTAURANT BASED ON RATING AND COST FOR TWO PEOPLE")
fig.show()
```
